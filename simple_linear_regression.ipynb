```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#read the file
df=pd.read_csv("/content/Salary_Data.csv")
df
```

```python
#assiging the varaiables to columns
#independent column must be in X (captial letter only)and 2D format
#y  is dependent column
X=df[["YearsExperience"]]
y=df["Salary"]
```

```python
import sklearn
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=3)
```

```python
len(X_train)
```

```python
len(y_train)
```

```python
len(X_test)
```

```python
len(y_test)
```

```python
#X_train in 2D format,y_train in 1D format, so we have to convert X_train 2D to 1D
training_data=pd.DataFrame({"X_train":X_train.to_numpy().ravel(),"y_train":y_train})
training_data
```

```python
#simple linear regression visualization
plt.figure(figsize=(5,3))
plt.title("simple linear regression")
plt.xlabel("Yearsof experience")
plt.ylabel("salary")
plt.scatter(x=X_train,y=y_train,color='r',marker='*')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression()
reg.fit(X_train,y_train)
```

```python
#m value
reg.coef_
```

```python
#c value
reg.intercept_
```

```python
#predicted answers
predicted_ans=reg.predict(X_train)
```

```python
#place the predicted_ans into the training_predicted_ans column
training_data["training_predicted_ans"]=predicted_ans
training_data
```

```python
#performing training performance by using r2_score for finding the accuracy
#r2_score is avialable in the sklearn.metrics
from sklearn.metrics import r2_score
r2_score(y_train,predicted_ans)
```

```python
#loss function by usig mean_squared_error
from sklearn.metrics import mean_squared_error
mean_squared_error(y_train,predicted_ans)
```

```python
plt.figure(figsize=(5,3))
plt.scatter(x=X_train,y=y_train,color='r',marker='*')
plt.plot(X_train,predicted_ans,color='black')
plt.show()
```
